## Functions: Default & Keyword Arguments

Functions are the building blocks of reusable code. Python supports default parameter values and keyword arguments, which make function calls more flexible and readable.

````python
def normalize(vec):
    """Return a list of values normalized to [0, 1]."""
    v_min = min(vec)
    v_max = max(vec)
    return [(x - v_min) / (v_max - v_min) for x in vec]

data = [3, 7, 1, 9, 4]
print(normalize(data))
````

````python
def stats(vec, decimals=4):
    """Compute mean and variance of a numeric list."""
    n = len(vec)
    mu = sum(vec) / n
    var = sum((x - mu) ** 2 for x in vec) / n
    return round(mu, decimals), round(var, decimals)

mu, var = stats([10, 20, 30, 40, 50])
print(f"Mean = {mu}, Variance = {var}")
````

## List/Tuple/Dict Basics

- **Lists** are mutable, ordered sequences.
- **Tuples** are immutable, ordered sequences.
- **Dictionaries** map keys to values.

````python
samples = [1.0, 2.5, 3.0, 4.5]  # list
coefficient = (0.97, -0.12)       # tuple
metadata = {"fs": 1e6, "n_samples": 1024}  # dict

print(f"List: {samples}, type={type(samples)}")
print(f"Tuple: {coefficient}, type={type(coefficient)}")
print(f"Dict: {metadata}, type={type(metadata)}")
print(f"Dict access: fs = {metadata['fs']}")
````

## For-Loops vs. List Comprehensions

````python
# For-loop approach
squares_loop = []
for x in range(10):
    squares_loop.append(x ** 2)

# List comprehension
squares_comp = [x ** 2 for x in range(10)]

print("Loop:      ", squares_loop)
print("Comprehension:", squares_comp)
````

````python
# Conditional comprehension: keep only even squares
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print("Even squares:", even_squares)
````

## Conditionals

````python
def classify_amplitude(amplitude, threshold=0.5):
    if amplitude > threshold:
        return "strong"
    elif amplitude > threshold / 2:
        return "moderate"
    else:
        return "weak"

for amp in [0.1, 0.3, 0.7]:
    print(f"Amplitude {amp} -> {classify_amplitude(amp)}")
````

## Reusable Helper Functions

Good signal-processing code relies on small, tested helper functions.

````python
def db_to_linear(db):
    """Convert decibels to linear scale."""
    import math
    return 10 ** (db / 20.0)

def linear_to_db(lin):
    """Convert linear magnitude to decibels."""
    import math
    if lin <= 0:
        return -120.0  # floor
    return 20.0 * math.log10(lin)

print(f"0 dB = {db_to_linear(0):.4f}")
print(f"6 dB = {db_to_linear(6):.4f}")
print(f"1.0 linear = {linear_to_db(1.0):.1f} dB")
print(f"2.0 linear = {linear_to_db(2.0):.1f} dB")
````

````python
def mean(vec):
    """Compute the arithmetic mean."""
    return sum(vec) / len(vec)

def variance(vec):
    """Compute the population variance."""
    m = mean(vec)
    return sum((x - m) ** 2 for x in vec) / len(vec)

def rms(vec):
    """Compute root-mean-square."""
    import math
    return math.sqrt(sum(x ** 2 for x in vec) / len(vec))

import random
random.seed(42)
test_signal = [random.gauss(0, 1) for _ in range(1000)]
print(f"Mean:   {mean(test_signal):.4f}")
print(f"Var:    {variance(test_signal):.4f}")
print(f"RMS:    {rms(test_signal):.4f}")
````

<details>
<summary>Solution</summary>

```python
def moving_average(data, window):
    return [sum(data[i:i+window]) / window for i in range(len(data) - window + 1)]

print(moving_average([1,2,3,4,5], 3))  # [2.0, 3.0, 4.0]
```
