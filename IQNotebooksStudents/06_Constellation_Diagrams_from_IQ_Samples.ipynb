# 06 - Constellation Diagrams from IQ Samples

This notebook generates constellation diagrams by scatter-plotting the real (I) vs. imaginary (Q) parts of IQ data.

````python
import numpy as np
import matplotlib.pyplot as plt
````

## What is a Constellation Diagram?

A constellation diagram plots IQ samples on a 2D plane:
- **X-axis**: I (real) component
- **Y-axis**: Q (imaginary) component

Each point represents one sample. Clusters of points correspond to modulation symbols. The spacing and spread of clusters reveal signal quality.

````python
# Generate dummy IQ data
np.random.seed(42)
n_samples = 4096
sample_rate = 1e6
t = np.arange(n_samples) / sample_rate

# BPSK-like: two clusters on the real axis
bits = np.random.choice([-1, 1], size=n_samples)
iq_bpsk = bits + 0j + 0.05 * (np.random.randn(n_samples) + 1j * np.random.randn(n_samples))

# QPSK-like: four clusters
symbols_i = np.random.choice([-1, 1], size=n_samples)
symbols_q = np.random.choice([-1, 1], size=n_samples)
iq_qpsk = symbols_i + 1j * symbols_q + 0.05 * (np.random.randn(n_samples) + 1j * np.random.randn(n_samples))
````

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.scatter(iq_bpsk.real, iq_bpsk.imag, s=1, alpha=0.3)
ax1.set_title('BPSK Constellation')
ax1.set_xlabel('I (Real)')
ax1.set_ylabel('Q (Imaginary)')
ax1.set_aspect('equal')
ax1.grid(True, alpha=0.3)
ax1.set_xlim(-2, 2)
ax1.set_ylim(-2, 2)

ax2.scatter(iq_qpsk.real, iq_qpsk.imag, s=1, alpha=0.3)
ax2.set_title('QPSK Constellation')
ax2.set_xlabel('I (Real)')
ax2.set_ylabel('Q (Imaginary)')
ax2.set_aspect('equal')
ax2.grid(True, alpha=0.3)
ax2.set_xlim(-2, 2)
ax2.set_ylim(-2, 2)

plt.tight_layout()
plt.show()
````

## Downsampling for Dense Constellations

When there are too many points, downsampling improves readability.

````python
step = 10
iq_downsampled = iq_qpsk[::step]

plt.figure(figsize=(6, 6))
plt.scatter(iq_downsampled.real, iq_downsampled.imag, s=5, alpha=0.5)
plt.title(f'QPSK Constellation (every {step}th sample)')
plt.xlabel('I (Real)')
plt.ylabel('Q (Imaginary)')
plt_aspect_equal = True
plt.gca().set_aspect('equal')
plt.grid(True, alpha=0.3)
plt.xlim(-2, 2)
plt.ylim(-2, 2)
plt.tight_layout()
plt.show()
````

## Overlaying Reference Points

````python
# QPSK reference points
ref_i = np.array([-1, -1, 1, 1])
ref_q = np.array([-1, 1, -1, 1])

plt.figure(figsize=(6, 6))
plt.scatter(iq_downsampled.real, iq_downsampled.imag, s=5, alpha=0.3, label='Received')
plt.scatter(ref_i, ref_q, c='red', s=100, marker='x', linewidths=2, label='Reference')
plt.title('QPSK Constellation with Reference Points')
plt.xlabel('I (Real)')
plt.ylabel('Q (Imaginary)')
plt.gca().set_aspect('equal')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(-2, 2)
plt.ylim(-2, 2)
plt.tight_layout()
plt.show()
````

## What Distortion or Rotation Indicates

- **Spread clusters** = noise or interference
- **Rotated constellation** = frequency offset or phase error
- **Offset from origin** = DC bias in I or Q
- **Elliptical clusters** = I/Q gain imbalance
- **Compressed clusters** = signal attenuation

## Summary

- A constellation diagram is a scatter plot of I vs. Q components.
- Use `plt.scatter(I, Q)` with `set_aspect('equal')` for proper geometry.
- Downsampling improves readability for large datasets.
- Cluster shape and spread reveal signal quality and distortion type.
