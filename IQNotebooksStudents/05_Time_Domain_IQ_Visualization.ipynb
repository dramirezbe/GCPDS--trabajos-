# 05 - Time-Domain IQ Visualization

This notebook visualizes IQ data in the time domain using matplotlib.

````python
import numpy as np
import matplotlib.pyplot as plt
````

## Generate Dummy IQ Data

````python
np.random.seed(42)
n_samples = 2048
sample_rate = 1e6  # 1 MHz
t = np.arange(n_samples) / sample_rate  # time in seconds

# Synthetic IQ: single tone + noise
freq_offset = 100e3  # 100 kHz tone
iq = (1.0 + 0.5j) * np.exp(2j * np.pi * freq_offset * t) + 0.1 * (np.random.randn(n_samples) + 1j * np.random.randn(n_samples))

print(f"IQ shape: {iq.shape}, dtype: {iq.dtype}")
print(f"Duration: {t[-1]*1e6:.1f} us")
````

## Plotting I and Q Separately

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 5), sharex=True)

ax1.plot(t * 1e6, iq.real, 'b-', linewidth=0.5)
ax1.set_ylabel('I (Real)')
ax1.set_title('I Component')
ax1.grid(True, alpha=0.3)

ax2.plot(t * 1e6, iq.imag, 'r-', linewidth=0.5)
ax2.set_ylabel('Q (Imaginary)')
ax2.set_xlabel('Time (\u03bcs)')
ax2.set_title('Q Component')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Complex Magnitude and Instantaneous Phase

````python
magnitude = np.abs(iq)
phase = np.angle(iq, deg=True)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 5), sharex=True)

ax1.plot(t * 1e6, magnitude, 'g-', linewidth=0.5)
ax1.set_ylabel('|IQ|')
ax1.set_title('Magnitude')
ax1.grid(True, alpha=0.3)

ax2.plot(t * 1e6, phase, 'm-', linewidth=0.5)
ax2.set_ylabel('Phase (\u00b0)')
ax2.set_xlabel('Time (\u03bcs)')
ax2.set_title('Instantaneous Phase')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Combined I/Q Overlay Plot

````python
plt.figure(figsize=(12, 4))
plt.plot(t * 1e6, iq.real, 'b-', linewidth=0.5, label='I (Real)', alpha=0.8)
plt.plot(t * 1e6, iq.imag, 'r-', linewidth=0.5, label='Q (Imaginary)', alpha=0.8)
plt.xlabel('Time (\u03bcs)')
plt.ylabel('Amplitude')
plt.title('IQ Signal - Time Domain')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## Computing a Time Axis from Sample Rate

The time axis is computed as:
```
t = np.arange(n_samples) / sample_rate
```
This gives evenly spaced time points starting at 0.

````python
print(f"Sample rate: {sample_rate/1e6:.1f} MHz")
print(f"Sample period: {1/sample_rate*1e9:.1f} ns")
print(f"Number of samples: {n_samples}")
print(f"Total duration: {n_samples/sample_rate*1e6:.1f} us")
print(f"Time step: {t[1]-t[0]:.2e} s")
````
